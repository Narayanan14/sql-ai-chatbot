```python
import os
import pandas as pd
from sqlalchemy import create_engine,text
from dotenv import load_dotenv
```

```python
pd.set_option('display.max_columns', None)
```

```python
load_dotenv()
```

```python
username=os.getenv("SNOWFLAKE_USER_NAME")
password=os.getenv("SNOWFLAKE_PASSWORD")
account=os.getenv("SNOWFLAKE_ACCOUNT")
warehouse=os.getenv("SNOWFLAKE_WAREHOUSE")
```

```python
try:
    engine = create_engine(f'snowflake://{username}:{password}@{account}?warehouse={warehouse}')
    conn=engine.connect()
    print('Connection established successfully')
except Exception as Error:
    print('An Error occured : {}'.format(Error))
```

```python
customer_master_df = pd.read_csv('Datasets/customer_master.csv')
```

```python
customer_master_df.info()
```

```python
customer_master_df.head()
```

```python
try:
    records_loaded=customer_master_df.to_sql('CUSTOMER_MASTER',con=conn,index=False,if_exists='append',schema='DATA_ANALYTICS.ECOMMERCEMART')
    print('Data Loaded Successfully')
    print('Number of Records Loaded : {}'.format(records_loaded))
except Exception as Error:
    print('An Error occured : {}'.format(Error))
```
